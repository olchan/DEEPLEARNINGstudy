> **빈도주의 vs 베이즈 관점**
- 빈도주의는 실험을 여러 번 했을 때 수렴하는 확률값이 각 사건의 확률이라고 생각

- P(A|B) = P(B|A) * P(A) / P(B)
- P(A) : A의 사전 확률 -> 이미 알고 있는 지식
- P(A|B) : A의 사후 확률
- P(B) : B의 사전 확률
- P(B|A) : 우도 likelihood -> 보통 알려져있음.
- 베이즈 정리는 관찰을 통해 얻은 새로운 정보를 활용하여 사후확률을 업데이트 한다는 개념

> Cross Entropy
- KL Divergence 이론은 어떤 data를 표현 가능한 최소 비트 수에 대한 정보량 개념에서, 이론적인 최소값과 실질적인 예측 확률을 통한 비트 수의 차이가 존재한다는 것을 말한다.
- 이떄 정답 p와 예측 확률 q로 구헤지는 비트 수를 최소화하면, 이론적인 최소값과의 차이가 줄어드므로, p와 q를 통해 구해지는 cross entropy를 최소화하는 하는 것이 cross entropy loss를 사용하는 이유이며, 정답 p는 train data의 label로 근사하여 사용하기에 지도학습에서의 손실함수로 사용된다.
